```python
#Cell 1 → Imports + Load Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

combined_df = pd.read_csv(
    "../data/processed/feature_engineered_data.csv"
)
```

```python
#Cell 2 → Basic Info
print(combined_df.head())

print(combined_df.info())

print(combined_df.describe())
```

```python
#Cell 3 → Missing Values
print(combined_df.isnull().sum())
```

```python
#Cell 4 → Revenue Distribution
plt.figure(figsize=(10,5))

sns.histplot(
    combined_df['revenue'],
    kde=True
)

plt.title("Revenue Distribution")

plt.show()
```

```python
#Cell 5 → ROI Distribution
plt.figure(figsize=(10,5))

sns.boxplot(
    x=combined_df['roi']
)

plt.title("ROI Distribution")

plt.show()
```

```python
#Cell 6 → Correlation Heatmap
plt.figure(figsize=(14,10))

sns.heatmap(
    combined_df.corr(numeric_only=True),
    annot=False,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
#Cell 7 Branch wise details
plt.figure(figsize=(10,5))

sns.barplot(
    x='brand',
    y='revenue',
    data=combined_df
)

plt.title("Brand-wise Revenue")

plt.show()
```
